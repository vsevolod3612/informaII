## Задание

Необходимо вычислить значение выражения:
$$Zze=\frac{35}{f} + y\cdot f - \frac{f+y}{4}$$
Для каждой переменной необходимо вывести ее значение и тип.


## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $f,y$. Переменная, которая должна быть вычислена на основе независимых: $Zze$.

```python
f=2
y=3
print("f = ", f, ", type(f) = ", type(f))
print("y =", y, ", type(y) = ", type(y))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Zze$:

```python
f = float(input("f: "))
y = float(input("y: "))
if f == 0:
 print("f не = 0")
else:
 zze = 35 / f + y * f - (f + y) / 4
 print(f"Zze = {zze}")
```
