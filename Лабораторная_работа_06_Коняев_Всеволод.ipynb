Теперь ответим на поставленный вопрос, используя строку `mystr`

````python
mystr = open('pr6_1.txt').read().strip()

count = 0

for i in range(1, len(mystr)):
  if mystr[i] == 'Q' and mystr[i - 1].isdigit():
    count += 1

print(f'{count}')
````
