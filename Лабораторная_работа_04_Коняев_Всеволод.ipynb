## Решение задачи

Заглавные гласные буквы (A, E, I, O, U) не могут входить в искомую подпоследовательность, поэтому заменим их на символ `*` и разрежем строку по этому символу. Получим части строки, в которых заглавных гласных нет.

В каждой части будем искать самую длинную подпоследовательность с четной суммой цифр. Для этого при обходе части накапливаем сумму цифр `total` и запоминаем в словаре `first`, где впервые встретилась четная и где впервые встретилась нечетная сумма. Если в двух позициях четность суммы одинакова, то сумма цифр между ними четная, а значит, подходящая подпоследовательность начинается после первой такой позиции и заканчивается на текущей.

```python
t = s
for v in 'AEIOU':
  t = t.replace(v, '*')

parts = t.split('*')

maxx = 0
maxs = ''

for part in parts:
  first = {0: 0}
  total = 0
  for j in range(len(part)):
    c = part[j]
    if c.isdigit():
      total += int(c)
    p = total % 2
    if p not in first:
      first[p] = j + 1
    length = j + 1 - first[p]
    if length > maxx:
      maxx = length
      maxs = part[first[p]:j + 1]

print(maxx)
print(maxs)
```

В данной работе были использованы операторы `if` (условный оператор), `for ... in` и `for ... in range(...)` (операторы цикла). Применялись функции `print()`, `len()`, `int()`, методы строк `replace()`, `split()`, `isdigit()`. В работе использовался тип данных `dict()`
