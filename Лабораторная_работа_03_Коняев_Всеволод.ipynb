## Задание

Напишите функцию, которая возвращает `True`, если переданное ей число является кубом некоторого натурального числа, в противном случае — `False`.

## Решение

### Объявление функции

Для проверки перебираем натуральные числа, пока их куб не станет больше проверяемого числа. Если куб совпал с исходным числом, функция возвращает `True`. Если совпадения нет, возвращается `False`. Натуральными считаются числа от 1 и выше, поэтому ноль и отрицательные числа не являются кубами натуральных чисел.

````python
def is_cube_of_natural(n):
    if n < 1:
        return False

    i = 1
    while i ** 3 <= n:
        if i ** 3 == n:
            return True
        i += 1

    return False
````

### Демонстрация использования функции

````python
numbers = [1, 8, 10, 27, 64, 100, 125, 0, -8]

for number in numbers:
    print(f"Число {number}: {is_cube_of_natural(number)}")
````

### Пример результата

При выполнении кода будет получен следующий результат:

```text
Число 1: True
Число 8: True
Число 10: False
Число 27: True
Число 64: True
Число 100: False
Число 125: True
Число 0: False
Число -8: False
```

## Ответы на контрольные вопросы

**1. Чем отличается функция с возвратом значения от функции без возврата значения?**

Функция с возвратом значения использует оператор `return` и передаёт результат туда, где её вызвали. Функция без возвращаемого значения обычно выполняет действие, например выводит информацию на экран, и не передаёт вызывающему коду полезный результат.

**2. Что такое библиотечная функция?**

Библиотечная функция — это готовая функция, предоставляемая Python или подключаемой библиотекой для выполнения определённой задачи.

**3. Почему библиотечные функции похожи на «чёрные ящики»?**

Потому что при использовании достаточно знать назначение функции, какие аргументы ей передать и какой результат она возвращает. Внутреннее устройство функции обычно не требуется изучать.

**4. Что делает инструкция `x = random.randint(1, 100)`?**

Она присваивает переменной `x` случайное целое число от 1 до 100 включительно. Для использования `random` предварительно нужно написать `import random`.

**5. Что делает инструкция `print(random.randint(1, 20))`?**

Она генерирует и выводит случайное целое число от 1 до 20 включительно. Вызов функции должен быть записан с круглыми скобками.

**6. Для чего используется `random.seed(20)`?**

Она задаёт начальное значение генератора псевдослучайных чисел. При одинаковом начальном значении и одинаковой последовательности вызовов генератор воспроизводит ту же последовательность результатов.

**7. Каким образом функции помогают повторно использовать код в программе?**

Функцию можно объявить один раз, а затем вызывать в разных местах программы. Это уменьшает повторение кода и упрощает его изменение и проверку.

**8. Назовите и опишите две части определения функции.**

Первая часть — заголовок функции: ключевое слово `def`, имя функции и параметры в скобках. Вторая часть — тело функции: блок команд с отступом, которые выполняются при вызове функции.

**9. Как у функции задать параметры по умолчанию?**

Нужно присвоить параметру значение в заголовке функции, например: `def greet(name="Гость"):`. Если аргумент не передан, используется значение по умолчанию.

**10. Что такое рекурсия?**

Рекурсия — это способ решения задачи, при котором функция вызывает саму себя. Для завершения рекурсивных вызовов необходимо предусмотреть условие остановки.
