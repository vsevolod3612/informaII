## Задание 2. Средний уровень — вариант 20
Даны `a = 5`, `b = 3`, `c = 2`. Вычислить выражение `a + b * c - a // c`, вывести результат и тип. Порядок вычислений пояснить в комментарии.

````python
a = 5
b = 3
c = 2

# Сначала выполняется умножение: b * c = 3 * 2 = 6.
# Затем целочисленное деление: a // c = 5 // 2 = 2.
# После этого сложение и вычитание: 5 + 6 - 2 = 9.
result = a + b * c - a // c

print("Результат:", result)
print("Тип результата:", type(result))
````

**Пояснение.** Оператор `*` и оператор `//` имеют более высокий приоритет, чем `+` и `-`. Целочисленное деление `5 // 2` возвращает `2`. Все операнды здесь целые числа, поэтому итоговый тип — `int`.

## Задание 3. Продвинутый уровень — вариант 20
Даны часы, минуты и секунды. Перевести время в общее количество секунд, затем выполнить обратный перевод в часы, минуты и секунды. Проверить, что исходные значения совпадают с результатом обратного перевода.

````python
hours = 2
minutes = 35
seconds = 48

total_seconds = hours * 3600 + minutes * 60 + seconds

# Обратный перевод: сначала находим полные часы,
# затем остаток минут и оставшиеся секунды.
hours_back = total_seconds // 3600
minutes_back = (total_seconds % 3600) // 60
seconds_back = total_seconds % 60

print(f"Исходное время: {hours} ч {minutes} мин {seconds} с")
print(f"Всего секунд: {total_seconds}")
print(f"После обратного перевода: {hours_back} ч {minutes_back} мин {seconds_back} с")

same = (hours == hours_back and
        minutes == minutes_back and
        seconds == seconds_back)
print("Исходные значения совпали с результатом:", same)
````

**Пояснение.** Один час равен 3600 секундам, одна минута — 60 секундам. Оператор `//` позволяет получить целое количество часов или минут, а `%` — остаток после деления. В конце сравниваются исходные значения и восстановленные.

## Контрольные вопросы

### 1. Что такое переменная в Python? Приведите корректные и некорректное имя.
Переменная — это имя, с которым связывается некоторое значение. Присваивание выполняется оператором `=`.

```python
age = 18
user_name = "Сева"
score2 = 95
```

Пример некорректного имени: `2score = 95`. Имя не может начинаться с цифры. Также нельзя использовать пробелы и ключевые слова Python в качестве имён переменных.

### 2. Основные простейшие типы данных и динамическая типизация.
- `int`: `5`, результат `type(5)` — `<class 'int'>`.
- `float`: `5.0`, результат — `<class 'float'>`.
- `str`: `"Привет"`, результат — `<class 'str'>`.
- `bool`: `True`, результат — `<class 'bool'>`.

Тип динамический, потому что тип определяется по присвоенному объекту и не требуется указывать его заранее. Одна переменная в разное время может ссылаться на значения разных типов.

### 3. Как работает `print()`?
Функция `print()` выводит данные на экран.

```python
name = "Сева"
print(name)
print("Имя:", name)
print(f"Меня зовут {name}")
```

В первом случае выводится одна переменная, во втором аргументы разделяются пробелом автоматически, в третьем используется форматированная строка.

### 4. Что делает `type()`?
`type()` возвращает тип объекта.

```python
number = 12
word = "текст"
flag = False

print(type(number))
print(type(word))
print(type(flag))
```

Результаты показывают соответственно типы `int`, `str` и `bool`.

### 5. Какие арифметические операции поддерживаются?
```python
print(7 + 2)   # 9, int
print(7 - 2)   # 5, int
print(7 * 2)   # 14, int
print(7 / 2)   # 3.5, float
print(7 // 2)  # 3, int
print(7 % 2)   # 1, int
print(7 ** 2)  # 49, int
print(7 + 2.0) # 9.0, float
```
Операторы: `+` — сложение, `-` — вычитание, `*` — умножение, `/` — обычное деление, `//` — целочисленное деление вниз, `%` — остаток, `**` — возведение в степень. Обычное деление возвращает `float`.

### 6. Разница между `/` и `//`.
`/` выполняет обычное деление, а `//` — деление с округлением результата вниз.

```python
print(7 / 2)    # 3.5
print(7 // 2)   # 3
print(-7 / 2)   # -3.5
print(-7 // 2)  # -4
```
Для отрицательных чисел `//` округляет вниз, то есть в сторону меньшего целого, а не просто отбрасывает дробную часть.

### 7. Преобразование типов.
`int()` преобразует значение в целое число, `float()` — в число с плавающей точкой, `str()` — в строку.

```python
print(int("123") + 1)   # 124
print(float("3.5"))     # 3.5
print(str(25))          # "25"
```
`int("abc")` вызывает `ValueError`, поскольку строка не представляет целое число. В пользовательских программах такие преобразования можно помещать в `try/except`.

### 8. Выражение, возвращающее значение.
Выражение вычисляется и даёт результат.

```python
total = 3 + 4
is_equal = (5 == 5)
is_allowed = (10 > 0 and 10 < 20)
print(total, is_equal, is_allowed)
```
Результаты: число `7`, логическое значение `True` и логическое значение `True`.

### 9. Операторы сравнения и логические операторы.
`==` — равно, `!=` — не равно, `<` — меньше, `>` — больше, `<=` — меньше или равно, `>=` — больше или равно. `and` означает «и», `or` — «или», `not` — отрицание.

```python
print(5 > 2 and 5 < 10)  # True
print(5 == 3 or 5 == 5)  # True
print(not (4 < 2))       # True
```
Сначала вычисляются арифметические операции, затем сравнения, затем логические операции с учётом их приоритета. Скобки помогают сделать порядок очевидным.

### 10. Операции над разными типами.
Числа `int` и `float` можно складывать: `5 + 2.0` даёт `7.0` типа `float`. Строку и число нельзя напрямую складывать: `"Возраст: " + 18` вызовет `TypeError`. Нужно явно преобразовать число: `"Возраст: " + str(18)` или использовать f-строку.

### 11. Способы вывода нескольких значений.
```python
name = "Сева"
age = 18

print(name, age)                  # через запятую
print(f"{name}, возраст: {age}")  # f-строка
print("{}: {}".format(name, age)) # format()
print(name + ": " + str(age))     # конкатенация
```
Запятая удобна для быстрого вывода разных типов, f-строки — для читаемого форматирования, `format()` — для шаблонов, конкатенация — для соединения строк после преобразования значений в строки.

### 12. Порядок выполнения операций.
Сначала выполняются операции с более высоким приоритетом: скобки, степени, умножение и деление, затем сложение и вычитание.

```python
print(2 + 3 * 4)    # 14
print((2 + 3) * 4)  # 20
print(10 - 6 // 2)  # 7
```
Скобки меняют порядок вычисления и повышают понятность кода.

### 13. Множественное присваивание.
Можно присвоить несколько значений одновременно, обменять значения или распаковать последовательность.

```python
a, b, c = 1, 2, 3
a, b = b, a
x, y = [10, 20]
print(a, b, c, x, y)
```

### 14. Изменяемые и неизменяемые типы.
Числа и строки относятся к неизменяемым типам: их значение нельзя изменить внутри существующего объекта. Например, операция `text = text + "!"` создаёт новую строку и связывает имя `text` с ней.

```python
text = "Привет"
old_id = id(text)
text = text + "!"
print(text)
print(old_id != id(text))
```
Сравнение `id()` обычно показывает, что после создания новой строки переменная ссылается на другой объект.

a = 7
b = 2.5
name = "Python"
total = a + b

`a` — целое число `int`; `b` — дробное число `float`; `name` — строка `str`; при сложении `int` и `float` получается `float`; `print()` выводит значения; `type()` показывает тип каждого объекта.
